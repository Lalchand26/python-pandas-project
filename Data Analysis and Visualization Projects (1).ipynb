```python
df['listed_in(type)'].value_counts()
```

```python
#5.Restaurent Showing in bar graph
df['listed_in(type)'].value_counts().plot(kind='bar')
plt.title("Different Restaurant Types")
plt.xlabel("Restaurant Type")
plt.ylabel("Count")

plt.show()
#Conclusion: The majority of the restaurants fall into the dining category.
```

```python
restaurent_sum=df.groupby('listed_in(type)')['votes'].sum()
result=pd.DataFrame({'vote':restaurent_sum})
plt.plot(result,c='green',marker='o')
plt.title('Different restaurent by line chart')
plt.xlabel('restaurent type')
plt.ylabel('vote')
plt.show()
```

```python
#Identify the Most Voted Restaurant
df.sort_values(by='votes',ascending=False).head(1)
```

```python
#Online Order Availability
df.groupby('online_order')['online_order'].count()
```

```python
pd.crosstab(df['online_order'], df['listed_in(type)'])
```

```python
df['online_order'].value_counts().plot(kind='bar',color='green')
plt.title('restaurent online availability')
plt.xlabel('online order')
plt.ylabel('restaurent data')
plt.show()
```

```python
#Analyze the approx_cost(for two people) column to find the preferred price range.
df['approx_cost(for two people)'].value_counts().head(10).plot(kind='bar')
plt.title('analysing the prefered price range')
plt.xlabel('price range')
plt.ylabel('number of restaurent')
plt.show()
#Conclusion: The majority of couples prefer restaurants with an approximate cost of 300 rupees.
```

```python
pd.crosstab(df['listed_in(type)'], df['online_order']).plot(
    kind='bar',stacked=True
)
plt.title('Order Mode Preferences by Restaurant Type')
plt.xlabel('restaurent type')
plt.ylabel('count')
plt.show()
# Dining restaurent prefered offline whereas cafe prefered online orders.
```
